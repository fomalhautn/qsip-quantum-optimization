## A short glossary before you begin

| Term | What it means here |
|---|---|
| **Decision variable** | One choice we are free to make, such as assigning a node to group 0 or 1 |
| **Objective** | The numerical score we want to maximize or minimize |
| **QUBO** | A *formulation*: a quadratic function of 0/1 choices (not a solver) |
| **Ising** | A mathematically equivalent formulation using -1/+1 choices |
| **Hamiltonian** | The quantum operator encoding the objective's score for each bitstring |
| **Qubit** | A quantum two-level system; its final measurement gives a classical 0 or 1 |
| **Tunable quantum circuit** | A circuit with adjustable rotation angles such as $\gamma$ and $\beta$ |
| **QAOA** | A *solver*: alternate cost and mixer operations; choose the angles using classical optimization |
| **Shot** | One execution and measurement of a quantum circuit; gives one candidate bitstring |
| **Expected score** | Average score over the circuit's output probabilities |

**How these fit:** Problem $\rightarrow$ QUBO/Ising formulation $\rightarrow$ QAOA circuit $\rightarrow$ simulated samples $\rightarrow$ compare against a classical optimum.
